# What moves the cascade boundary?
Run from the repository root, using its Python environment. The shock draws are shared across each comparison.

```python
from study import analyze, simulate
from random import Random
rng = Random(8808)
shocks = [.20*rng.betavariate(2,5) for _ in range(400)]
for impact in (.08,.16,.32):
    for cap in (0,30,60,120):
        row = analyze(shocks, 8, cap, impact=impact)
        print(f'impact={impact:.2f}, capacity={cap}: {row["cascade"]:.1%} ± {row["cascade_mcse"]:.1%} MC SE')
```
